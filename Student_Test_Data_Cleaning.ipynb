### Step 2: Standardize Column Names
Rename columns to snake_case:
`student_id`, `student_name`, `email`, `age`, `gender`, `city`, `department`, `course`, `marks`, `annual_salary`, `phone`, `admission_date`

```python
# TODO: Rename columns
df.rename(columns={
    'Student ID': 'student_id',
    'Full Name': 'student_name',
    'Contact Email': 'email',
    'Age': 'age',
    'Gender': 'gender',
    "City": "city",
    "Department": "department",
    "Degree Course": "course",
    "Marks Percentage": "marks",
    "Annual Package": "annual_salary",
    "Mobile Number": "phone",
    "Enrollment Date": "admission_date"
}, inplace=True)
df.columns
```

### Step 3: Handle Missing Values & Placeholders
- Replace placeholders (`'NA'`, `'N/A'`, `'null'`, `'Not Available'`, `'unknown'`) with `np.nan`.
- Impute categorical columns with `'Unknown'`.
- Impute missing numeric values using column medians.

```python
missing_count = df.isna().sum()
print(missing_count)
# TODO: Replace string placeholders with NaN
missing_values = ['NA', 'N/A', 'null', 'Not Available', 'unknown']
df.replace(missing_values, np.nan , inplace=True)

for column in df.columns:
    if df[column].dtype == 'object':
        df[column].fillna('Unknown', inplace=True)
# TODO: Impute categorical & numeric missing values
df["age"] = pd.to_numeric(df["age"], errors='coerce')
df["annual_salary"] = pd.to_numeric(df["annual_salary"], errors='coerce')
df["marks"] = pd.to_numeric(df["marks"], errors='coerce')
catagorical_columns = ["student_name", "email", "course"]
numeric_columns = ["age", "annual_salary", "marks","gender"]

df["age"].fillna(df["age"].median(), inplace=True)
df["annual_salary"].fillna(df["annual_salary"].median(), inplace=True)
df["marks"].fillna(df["marks"].median(), inplace=True)

```

### Step 4: Text Cleaning & Category Standardization
- Strip whitespaces.
- Standardize: `bangalore` -> `bengaluru`, `gurgaon` -> `gurugram`.
- Standardize departments (`computer science` -> `cse`, etc.).
- Standardize degree courses (`btech`, `mtech`, `bca`, `bsc`).

```python
df["course"].unique()
```

```python
# TODO: Clean and standardize text columns
df["city"] = df["city"].str.strip()
df["city"] = df["city"].replace({
    "bangalore": "bengaluru",
    "gurgaon": "gurugram"})
df["department"] = df["department"].str.strip()
df["department"] = df["department"].str.lower()
df["department"] = df["department"].replace({
    "Computer Science":"cse",
    "Information Technology":"it",
    "Mechanaical Engineering":"me",
    "Data Science":"ds",
    "Electronics":"ece",
    "data science": "ds",
})

df["gender"] = df["gender"].str.strip()
df["gender"] = df["gender"].str.lower()
df["gender"] = df["gender"].replace({
    'Male':"m",
    'male':"m",
   'Female':"f",
    'other':"o",
    'FEMALE':"o",
    'female':"o",
   'Other':"o",
    'MALE':"m"

})

df["course"] = df["course"].str.strip()
df["course"] = df["course"].str.lower()
df["course"] = df["course"].replace({
    'MTECH':"mtech",
    'B.Sc':"bsc",
    'm.tech':"mtech",
    'bca':'bca',
    'BTECH':'btech',
    'B.Tech':'btech','BCA':'bca',
    'M.Tech':'btech',
    'b.tech':'btech',

})

```

### Step 5: Remove Duplicates
- Identify and drop duplicate rows.

```python
# TODO: Count and drop duplicate rows
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

### Step 6: Validate Numeric Boundaries & Outliers
- `age`: 17 to 35
- `marks`: 0 to 100
- `annual_salary`: > 0 and < 10,000,000

```python
# TODO: Fix invalid numeric ranges and fill with median
df=df[df['age'].between(17,35)]
df=df[df['marks'].between(0,100)]
df=df[df['annual_salary'].between(0,10000000)]
```

### Step 7: Email & Phone Validation
- Validate email pattern.
- Clean phone numbers (10 digits starting with 6-9).

```python
# TODO: Validate emails and clean mobile numbers
df["phone"] = df["phone"].astype("string").str.replace(r"\D","",regex=True)
invalid_phone = ~df["phone"].str.match(r"[6-9]\d{9}$",na=False)
print("Invalid phone:",invalid_phone.sum())
df.loc[invalid_phone , "phone"] = np.nan


df["email"]=df["email"].str.strip().str.lower()

invalid_email = ~df["email"].str.match(r"^[^\s@]+@[^\s@]+\.[^\s@]+$", na=False)
print("Invalid Emails:", invalid_email.sum())

df.loc[invalid_email, "Email_address"]=np.nan

```

### Step 8: Parse Dates & Export
- Convert `admission_date` to datetime.
- Export final cleaned DataFrame to `clean_student_data_80k.csv`.

```python
df["admission_date"] = pd.to_datetime(df["admission_date"],errors = "coerce",dayfirst = True)
clean_df = df.copy()
clean_df.info()
```
